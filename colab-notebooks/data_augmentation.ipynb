### Data Augmentation

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
# Load the Iris dataset
iris = load_iris()
#X - features
#y - labels
X, y = iris.data, iris.target
#50 c0,50- c1, 50-c2 # balanced
#imbalanced
#50 c0, 25-c1
# Choose two classes (0 and 1) and take 50 samples from class 0 and 25 samples from class 1
X_imbalanced = np.concatenate([X[y == 0][:50], X[y == 1][:25]])
y_imbalanced = np.concatenate([y[y == 0][:50], y[y == 1][:25]])
X_imbalanced
y_imbalanced
```

```python
# Display class distribution before addressing imbalance
print("Class Distribution Before Sampling:")
print(pd.Series(y_imbalanced).value_counts())
```

```python
# Apply Random Over-sampling
#c0 - majority 50
#c1 - minority 25
#over
#c0-50
#c1-50
from imblearn.over_sampling import RandomOverSampler
# Apply Random Over-sampling
ros = RandomOverSampler(random_state=42)
X_resampled, y_resampled = ros.fit_resample(X_imbalanced, y_imbalanced)
# Display class distribution after oversampling
print("\nClass Distribution After Random Over-sampling:")
print(pd.Series(y_resampled).value_counts())
```

```python
from imblearn.over_sampling import SMOTE
# Apply SMOTE
smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X_imbalanced, y_imbalanced)
# Display class distribution after SMOTE
print("\nClass Distribution After SMOTE:")
print(pd.Series(y_resampled).value_counts())
```
