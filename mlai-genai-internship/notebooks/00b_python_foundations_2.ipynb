# Concept 5 — Collections: list, tuple, set, dict

🧠 **Analogy: four kinds of container.**

| Type | Analogy | Ordered? | Changeable? | Duplicates? | Written as |
|---|---|---|---|---|---|
| **list** | A shelf of books | Yes | Yes | Yes | `[1, 2, 3]` |
| **tuple** | A sealed box of coordinates | Yes | **No** | Yes | `(1, 2)` |
| **set** | A bag of raffle tickets | No | Yes | **No** | `{1, 2, 3}` |
| **dict** | A labelled drawer | Yes | Yes | Keys unique | `{"a": 1}` |

**Nine times out of ten the answer is a list or a dict.**

- Use a **list** when order matters and you will add or remove things
- Use a **dict** when you want to look things up by name rather than position
- Use a **set** when you only care whether something is present, or want duplicates gone
- Use a **tuple** when the values must never change

## 📘 Examples

### Example 1 — lists: the workhorse

````python
marks = [78, 92, 65, 88, 45]

print("the list  :", marks)
print("length    :", len(marks))
print("first     :", marks[0])
print("last      :", marks[-1])
print("slice 1:3 :", marks[1:3], "  includes 1, EXCLUDES 3")

# Lists can be changed after they are made.
marks.append(97)                 # add to the end
print("\nafter append:", marks)

marks[0] = 80                    # change one item
print("after change:", marks)

marks.remove(45)                 # remove by VALUE
print("after remove:", marks)

# Useful built-ins that work on any list of numbers
print(f"\nsum {sum(marks)} | max {max(marks)} | min {min(marks)} "
      f"| average {sum(marks) / len(marks):.1f}")
print("sorted:", sorted(marks))
````

### Example 2 — dicts: look things up by name

````python
student = {"name": "Asha", "age": 21, "cgpa": 8.45}

print("the dict :", student)
print("by key   :", student["name"])

# .get() is safer: it returns None instead of crashing on a missing key.
print("missing with [] would CRASH; .get() gives:", student.get("phone"))
print("with a default            :", student.get("phone", "not provided"))

# Add or change
student["semester"] = 6
student["cgpa"] = 8.60
print("\nupdated:", student)

# Loop over a dict: .items() gives you key AND value
print()
for key, value in student.items():
    print(f"  {key:<10}: {value}")

print("\nall keys  :", list(student.keys()))
print("all values:", list(student.values()))
````

### Example 3 — sets and tuples, and when each earns its place

````python
# SET: duplicates vanish automatically. Perfect for "what distinct values exist?"
cities = ["Delhi", "Mumbai", "Delhi", "Chennai", "Mumbai", "Delhi"]
unique_cities = set(cities)

print("original:", cities)
print("as a set:", unique_cities)
print("how many distinct:", len(unique_cities))

# Sets answer "is it in there?" very fast, and do maths on groups.
python_students = {"Asha", "Ravi", "Meera"}
ml_students = {"Ravi", "Meera", "John"}

print("\nin BOTH courses  :", python_students & ml_students)
print("in EITHER course :", python_students | ml_students)
print("Python only      :", python_students - ml_students)

# TUPLE: fixed. Use it when the values must not change.
location = (10.85, 76.27)         # latitude, longitude
print("\nlocation:", location)
try:
    location[0] = 99
except TypeError as error:
    print("Cannot change a tuple:", error)
````

# Concept 6 — Conditions and loops

🧠 **Analogy: a recipe with decisions and repetition.**

**Conditions** are the "if the sauce is too thick, add water" step. **Loops** are the "stir for 20 minutes" step.

> ⚠️ **Indentation is not decoration in Python — it is the syntax.** The indented lines are the ones that belong to the `if` or the `for`. Get it wrong and the program does something different, often without any error.

## 📘 Examples

### Example 1 — if / elif / else

````python
marks = 78

if marks >= 90:
    grade = "A"
elif marks >= 75:          # elif = "otherwise, if"
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "F"

print(f"{marks} -> {grade}")

# WHY elif AND NOT A SECOND if:
print("\nUsing if twice by mistake:")
m = 95
g = ""
if m >= 90:
    g = "A"
if m >= 75:        # this ALSO runs, and overwrites the A
    g = "B"
print(f"  {m} -> {g}   <- WRONG, should be A")

print("\nUsing elif correctly:")
g = "A" if m >= 90 else ("B" if m >= 75 else "C")
print(f"  {m} -> {g}   <- correct")
````

### Example 2 — for loops

````python
marks = [78, 92, 65, 88, 45]

# Loop over the VALUES
for mark in marks:
    print("mark:", mark)

# enumerate() gives position AND value - no counter variable needed
print()
for index, mark in enumerate(marks):
    print(f"  Subject {index + 1}: {mark}")

# range() makes a sequence of numbers
print("\nrange(5)      :", list(range(5)))
print("range(1, 6)   :", list(range(1, 6)))
print("range(0, 10, 2):", list(range(0, 10, 2)))

# zip() walks two lists together
subjects = ["Maths", "Physics", "Python", "Stats", "English"]
print()
for subject, mark in zip(subjects, marks):
    status = "pass" if mark >= 60 else "FAIL"
    print(f"  {subject:<10} {mark:>3}  {status}")
````

### Example 3 — building up a result, and list comprehensions

````python
marks = [78, 92, 65, 88, 45]

# The pattern you will use constantly: start empty, add as you go.
passed = []
for mark in marks:
    if mark >= 60:
        passed.append(mark)
print("passed:", passed)

# A list comprehension does the same in one line.
passed_short = [m for m in marks if m >= 60]
print("same  :", passed_short)

# With a transformation too:
boosted = [m + 5 for m in marks]
print("boosted:", boosted)

# Counting with a running total
total = 0
for mark in marks:
    total = total + mark
print(f"\ntotal {total}, average {total / len(marks):.1f}")

# while: repeat until a condition stops being true
countdown = 3
while countdown > 0:
    print("countdown:", countdown)
    countdown = countdown - 1      # MUST change, or it loops forever
print("go!")
````

# Concept 7 — Functions

🧠 **Analogy: a recipe card.**

A function is a set of instructions you write **once**, give a name, and then use whenever you like. You hand it ingredients (arguments) and it hands back a dish (the return value).

**Why it matters:** if you write the same six lines in four places and later find a bug, you must fix it four times. In a function you fix it once.

```text
def   calculate_percentage  (obtained, total=100)  :
 ↑            ↑                  ↑         ↑
keyword     name            arguments   default
```

## 📘 Examples

### Example 1 — writing and calling a function

````python
# def creates the function. It does NOT run it.
def greet(name):
    """Print a greeting. The text in triple quotes is the docstring."""
    print(f"Hello, {name}!")

# Calling it is what runs it.
greet("Asha")
greet("Ravi")

# return HANDS A VALUE BACK, so you can use the result.
def calculate_percentage(obtained, total=100):
    """Return the percentage, rounded to 2 decimal places."""
    return round((obtained / total) * 100, 2)

result = calculate_percentage(87)
print("\nresult:", result)
print("with a different total:", calculate_percentage(87, total=120))

# print() SHOWS a value. return GIVES IT BACK so you can keep using it.
doubled = calculate_percentage(45) * 2
print("you can do maths with a returned value:", doubled)
````

### Example 2 — arguments, defaults and keyword arguments

````python
def student_report(name, marks, passing=40, show_grade=True):
    """Build a one-line report for a student."""
    status = "PASS" if marks >= passing else "FAIL"
    line = f"{name:<10} {marks:>3}  {status}"
    if show_grade:
        grade = "A" if marks >= 90 else "B" if marks >= 75 else "C" if marks >= 60 else "D"
        line += f"  grade {grade}"
    return line

# Positional: order matters
print(student_report("Asha", 87))

# Keyword: order does not matter, and it is far clearer
print(student_report(name="Ravi", marks=35))
print(student_report("Meera", 92, passing=50, show_grade=False))

# Defaults mean you only specify what differs from the usual case.
````

### Example 3 — functions that build on each other

````python
def average(numbers):
    """Return the average of a list of numbers."""
    if not numbers:                  # guard against an empty list
        return 0
    return sum(numbers) / len(numbers)

def grade_from(mark):
    """Return a letter grade for one mark."""
    if mark >= 90: return "A"
    if mark >= 75: return "B"
    if mark >= 60: return "C"
    return "F"

def class_report(students):
    """Print a report for a dict of {name: [marks]}."""
    all_averages = []
    for name, marks in students.items():
        avg = average(marks)
        all_averages.append(avg)
        print(f"  {name:<10} {avg:>6.1f}  {grade_from(avg)}")
    print(f"  {'CLASS':<10} {average(all_averages):>6.1f}")

class_report({
    "Asha":  [78, 92, 65],
    "Ravi":  [88, 74, 91],
    "Meera": [95, 89, 97],
})

print("\nEach function does ONE job. That is what makes them reusable.")
````

**Q1.** What is the difference between `print()` and `return` inside a function?

**Q2.** What does `def` do — does it run the function?

**Q3.** In `def f(x, total=100)`, what is `total=100` and why is it useful?

**Q4.** Why is `student_report(name="Ravi", marks=35)` often better than `student_report("Ravi", 35)`?

<details>
<summary><b>Answers</b></summary>

**A1.** `print()` **shows** something on the screen and the value is then gone. `return` **hands the value back** so the caller can store it, use it in maths, or pass it on. A function that only prints cannot be built upon.

**A2.** **No.** `def` only *defines* the function: it creates the recipe card and gives it a name. Nothing runs until you **call** it with `greet("Asha")`.

**A3.** A **default argument**. Callers may leave it out and it takes the value 100. It lets one function handle the common case simply while still allowing the unusual case.

**A4.** **Keyword arguments** are self-documenting and order-independent. With four or five arguments, positional calls become impossible to read and easy to get wrong.

**A5.** Put them in a function. Otherwise, when you find a bug you must fix it in four places — and you will miss one. **A function means one definition, one place to fix.**
