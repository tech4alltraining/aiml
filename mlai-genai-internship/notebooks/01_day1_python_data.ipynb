# Concept 1 — Python refresher

You need six things fluently: **variables, collections, indexing, conditionals, loops, functions.** Nothing more, for now.

🧠 **Analogy: the four containers.** Python's collections are four kinds of container.

| Type | Analogy | Ordered? | Changeable? | Duplicates? |
|---|---|---|---|---|
| **list** `[ ]` | A shelf of books | Yes | Yes | Yes |
| **tuple** `( )` | A sealed box of coordinates | Yes | **No** | Yes |
| **dict** `{k: v}` | A labelled drawer | Yes | Yes | Keys unique |
| **set** `{ }` | A bag of raffle tickets | No | Yes | **No** |

Nine times out of ten the answer is a **list** or a **dict**.

## 📘 Examples

Three worked examples. Run each, then read the output next to the code.

### Example 1 — variables, collections, indexing

````python
# Python works out the type for you; you never declare it.
name = "Asha"        # str
age = 21             # int
score = 87.5         # float
enrolled = True      # bool

print(name, age, score, enrolled)
print(type(name).__name__, type(age).__name__, type(score).__name__)

# The four collections
marks = [78, 92, 65, 88]                 # list  - ordered, changeable
point = (10.5, 20.3)                     # tuple - fixed once made
student = {"name": "Asha", "marks": 87}  # dict  - looked up by key
subjects = {"Math", "Physics", "Math"}   # set   - duplicates vanish

print("\nlist :", marks)
print("tuple:", point)
print("dict :", student)
print("set  :", subjects, " <- only ONE 'Math'")

# Indexing counts from 0. A slice INCLUDES start, EXCLUDES stop.
print("\nmarks[0]   =", marks[0],   " first item")
print("marks[-1]  =", marks[-1],  " last item")
print("marks[1:3] =", marks[1:3], " items 1 and 2, NOT 3")
````

### Example 2 — conditionals and loops

````python
average = sum(marks) / len(marks)

# elif means "otherwise, if". Using a second `if` is a classic bug:
# both branches could then run and the later one wins.
if average >= 90:
    grade = "A"
elif average >= 75:
    grade = "B"
else:
    grade = "C"

print(f"Average {average:.2f} -> grade {grade}")

# enumerate() gives position AND value, so you never need a counter variable.
print()
for index, mark in enumerate(marks):
    status = "pass" if mark >= 75 else "needs work"
    print(f"  Subject {index + 1}: {mark:>3}  ({status})")
````

### Example 3 — functions and list comprehensions

````python
def calculate_percentage(obtained, total=100):
    """Return the percentage, rounded to 2 places. total defaults to 100."""
    return round((obtained / total) * 100, 2)

print(calculate_percentage(87))             # uses the default
print(calculate_percentage(87, total=120))  # overrides it

# List comprehension: build a list in one line.
# You will READ this constantly in ML code, so learn to recognise it.
squared = [m ** 2 for m in marks]           # transform every item
passed  = [m for m in marks if m >= 75]     # keep only some items

print("\noriginal:", marks)
print("squared :", squared)
print("passed  :", passed)

# The same thing written the long way, for comparison:
passed_long = []
for m in marks:
    if m >= 75:
        passed_long.append(m)
print("same    :", passed_long)
````

# Concept 2 — NumPy: fast arrays

🧠 **Analogy: the shopping bag and the egg tray.**

A Python **list** is a shopping bag — throw anything in, but to act on everything you must pull items out one at a time.

A NumPy **array** is an egg tray — every slot holds the same kind of thing, in a fixed grid. Because the computer knows every slot is identical, it can act on **all of them at once**.

On a million numbers the egg tray is roughly fifty times faster. **That is the entire reason NumPy exists.**

## 📘 Examples

### Example 1 — the difference that matters

````python
python_list = [100, 200, 300]
numpy_array = np.array([100, 200, 300])

print("list  * 2 =", python_list * 2)
print("            ^ the BAG was duplicated - probably not what you wanted")
print()
print("array * 2 =", numpy_array * 2)
print("            ^ every ELEMENT was doubled - this is what we want")
````

### Example 2 — shapes, and maths on whole arrays

````python
a = np.array([1, 2, 3, 4, 5])
b = np.array([[1, 2, 3],
              [4, 5, 6]])

print("a       :", a)
print("a.shape :", a.shape, "  5 elements, 1 dimension")
print("b.shape :", b.shape, " 2 rows, 3 columns")
print("a.dtype :", a.dtype, "  every element is this type")

# Every operation applies to the WHOLE array. No loop anywhere.
print("\na + 10  :", a + 10)
print("a ** 2  :", a ** 2)
print("a / 2   :", a / 2)
````

### Example 3 — statistics and boolean masking

````python
data = np.array([23, 45, 12, 67, 34, 89, 21])

print("data  :", data)
print("mean  :", data.mean().round(2))
print("median:", np.median(data))
print("std   :", data.std().round(2))
print("min/max:", data.min(), data.max())

# BOOLEAN MASKING - the idea behind every Pandas filter you will ever write.
mask = data > 30                 # an array of True/False, one per element

print("\ndata > 30  :", mask)
print("data[mask] :", data[mask], " <- only the True positions")
print("how many   :", mask.sum(), "  (True counts as 1)")
````

# Concept 3 — Pandas: tables of data

🧠 **Analogy: a spreadsheet that takes orders.**

A `DataFrame` is a spreadsheet — rows, columns, headers. The difference is that instead of clicking and dragging, you give it instructions.

| What you would do in Excel | What you type in Pandas |
|---|---|
| Scroll to the top to check the data | `df.head()` |
| Read the row count off the status bar | `df.shape` |
| Filter → "Price greater than 100000" | `df[df["price"] > 100000]` |
| Sort descending by Price | `df.sort_values("price", ascending=False)` |
| Insert a PivotTable | `df.groupby("fuel")["price"].mean()` |
| Find blanks with conditional formatting | `df.isnull().sum()` |
| Add a formula column | `df["new"] = df["a"] / df["b"]` |

**You are not learning a new idea — you are learning new words for one you already have.**

## 📘 Examples

### Example 1 — load a dataset and look at it

````python
# 15,000 used cars with their selling prices.
df = pd.read_csv(BASE + "regression/cardekho_dataset.csv")

print("Rows and columns:", df.shape)
df.head()
````

### Example 2 — selecting and filtering

````python
# ONE name in [] gives a Series. A LIST of names gives a DataFrame.
print("one column ->", type(df["selling_price"]).__name__)
print("two columns ->", type(df[["brand", "selling_price"]]).__name__)

# FILTERING: use & for AND and | for OR, and wrap EACH condition in brackets.
# Plain `and` / `or` do not work on whole columns.
expensive = df[df["selling_price"] > 1000000]
print("\nCars above 10 lakh:", len(expensive))

recent_petrol = df[(df["vehicle_age"] < 5) & (df["fuel_type"] == "Petrol")]
print("Recent petrol cars:", len(recent_petrol))

recent_petrol[["car_name", "vehicle_age", "selling_price"]].head()
````

### Example 3 — grouping, sorting and new columns

````python
# GROUPING is the single most useful Pandas skill.
# "For each fuel type, what is the average price?"
print("Average price by fuel type:")
print(df.groupby("fuel_type")["selling_price"].mean().round(0).sort_values(ascending=False))

# A new column from existing ones.
df["price_lakhs"] = (df["selling_price"] / 100000).round(2)

# Sorting.
print("\nThree most expensive cars:")
print(df.sort_values("selling_price", ascending=False)
        [["car_name", "price_lakhs", "fuel_type"]].head(3))

# Counting categories.
print("\nHow many of each fuel type:")
print(df["fuel_type"].value_counts())
````

# Concept 4 — Exploratory Data Analysis

🧠 **Analogy: the doctor's check-up.**

A doctor does not prescribe medicine the moment you walk in. First comes the check-up: temperature, blood pressure, a few questions. Only then a treatment.

**EDA is the check-up for your dataset.** Students who skip it and jump straight to `model.fit()` are prescribing medicine without examining the patient — and then wondering why it did not work.

### The five questions, in this order

1. **How big is it?** `df.shape`
2. **What is each column?** `df.info()`
3. **What is missing?** `df.isnull().sum()`
4. **What are the ranges and outliers?** `df.describe()`
5. **What is the target, and is it balanced?** `df["target"].value_counts()`

## 📘 Examples

### 📊 How to read a `describe()` table

| Row | What it tells you | The check to make |
|---|---|---|
| `count` | How many **non-missing** values | Lower than the row count? You have gaps in this column |
| `mean` | The average | Compare with `50%` — a big gap means outliers are dragging it |
| `std` | How spread out the values are | Larger than the mean? Very wide spread, probably outliers |
| `min` | Smallest value | Negative where it should not be? A data error |
| `25%` / `50%` / `75%` | The quartiles. `50%` is the **median** | These are where the bulk of your data actually lives |
| `max` | Largest value | **Many times bigger than `75%`? That is an outlier** |

**Applied to the Salary column above:**

- `75%` is **81,000** but `max` is **1,500,000** → the outlier, now measured rather than eyeballed
- `mean` is about **196,091** but the median is **67,000** → one row dragged the average up by nearly 130,000

> **That second point is exactly why you fill missing numbers with the median, not the mean.** You have now seen it happen rather than been told it.

### Example 3 — the full five-question sweep on a real dataset

````python
loans = pd.read_csv(BASE + "loan_data_10k.csv")

print("=== 1. How big is it? ===")
print(loans.shape)

print("\n=== 3. What is missing? ===")
missing = loans.isnull().sum()
print(missing[missing > 0])

print("\n=== 5. Is the target balanced? ===")
print(loans["loan_status"].value_counts())
print(loans["loan_status"].value_counts(normalize=True).round(3))
````

````python
# Which numeric columns relate to the target?
# Values near +1 or -1 move WITH the target. Near 0 means no LINEAR
# relationship on its own.

loans.corr(numeric_only=True)["loan_status"].sort_values().round(3)
````

### 📊 How to read a correlation column

| Value you see | What it means |
|---|---|
| **+0.7 to +1.0** | Strong positive — as this column rises, the target rises |
| **+0.3 to +0.7** | Moderate positive |
| **−0.3 to +0.3** | Weak *linear* relationship. **Not the same as "useless"** — the relationship may be curved, or only matter alongside another column |
| **−0.7 to −0.3** | Moderate negative |
| **−1.0 to −0.7** | Strong negative — as this column rises, the target falls |

**Two warnings students routinely trip over:**

1. **A correlation of −0.3 is not "weak" in real data.** In messy human data it is worth investigating. Do not discard a column because it scored 0.2.
2. **Correlation is not causation.** Ice cream sales correlate with drowning deaths. Neither causes the other — hot weather causes both.

> **In one sentence:** this column tells you which features are *worth a closer look*, not which features to keep.

## 🧠 Quick quiz — EDA

Answer all five from memory. No scrolling back.

**Q1.** Why do EDA at all, instead of going straight to `model.fit()`?

**Q2.** In a `describe()` output, `mean` is 196,091 and `50%` is 67,000. What does that gap tell you?

**Q3.** Your target is 92% class 0 and 8% class 1. What does that mean for your choice of metric?

**Q4.** Two columns correlate at 0.85. Does one cause the other?

**Q5.** A column correlates with the target at 0.15. Should you delete it?

<details>
<summary><b>Answers</b></summary>

**A1.** Because a model cannot tell you that your data is broken — it will happily train on duplicated rows, mis-scaled units and leaked columns, and give you a confident wrong answer. EDA is the check-up before the prescription.

**A2.** The mean is far above the median, so a small number of very large values are dragging the average up. There are outliers in this column, and the median is the more honest summary.

**A3.** The classes are imbalanced, so accuracy is misleading — always predicting class 0 scores 92% and catches nothing. Report recall, precision and F1 for the minority class instead.

**A4.** No. Correlation measures that they move together, not why. A third factor may drive both, or the link may be coincidence. Causation needs an experiment or domain knowledge, not a correlation number.

**A5.** Not on that basis. Correlation only measures **linear** relationship on its own. The column may matter in combination with another, or its relationship may be curved. Test it in a model before discarding it.
