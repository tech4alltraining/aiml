# Concept 1 — Open-source models vs API models

🧠 **Analogy: the taxi and the car.**

An **API model** (Gemini, GPT, Claude) is a **taxi** — you pay per trip, someone else maintains the vehicle, and you can start using it in five minutes.

An **open-source model** (Llama, Mistral, Qwen, Gemma) is **buying a car** — free to drive once you own it, but you need somewhere to park it and you do the maintenance.

| | API model | Open-source model |
|---|---|---|
| Where it runs | The provider's servers | Your machine or server |
| Cost | Per token | Free weights; you pay for hardware |
| Setup | An API key | Download weights, enough RAM/GPU |
| Data privacy | Data leaves your network | Data can stay entirely local |
| Customisation | Prompting only | Full fine-tuning possible |
| Best for | Getting started, best quality, no infrastructure | Privacy, cost at scale, offline use |

## 📘 Examples

### Example 1 — will this model even fit on your machine?

````python
# The number in a model's name (7B, 13B, 70B) is its PARAMETER COUNT
# in billions. Rough rule: RAM needed ~ parameters x bytes-per-parameter.

models = [
    ("distilgpt2",        0.082),
    ("gpt2",              0.124),
    ("Mistral-7B",        7.0),
    ("Llama-3-8B",        8.0),
    ("Llama-3-70B",      70.0),
]

print(f"{'model':<18}{'params(B)':>11}{'float32':>10}{'float16':>10}{'4-bit':>9}")
print("-" * 58)
for name, params in models:
    print(f"{name:<18}{params:>11.3f}"
          f"{params * 4:>9.1f}G{params * 2:>9.1f}G{params * 0.5:>8.1f}G")

print("\nA typical laptop has 8-16 GB of RAM.")
print("Check the parameter count BEFORE you try to download something.")
````

# Concept 2 — The `pipeline()` function

🧠 **Analogy: the app store for AI models.**

Nobody writes their own camera app. You open the app store, search, read the reviews, check the size, and install one.

**Hugging Face is that for AI models** — over a million of them, free, searchable by task.

`pipeline()` handles tokenising, running the model and decoding the output in one call. It is the fastest way to use any model on the Hub.

## 📘 Examples

### Example 1 — sentiment analysis

````python
from transformers import pipeline

# ALWAYS name the model explicitly.
#
# pipeline("sentiment-analysis") with no model gives you whatever default
# the library currently ships. That default CHANGES between versions, so
# your results stop being reproducible - and a marker re-running your
# notebook next month may get different numbers.

sentiment = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english",
)

sentences = [
    "I love machine learning.",
    "This movie was very boring.",
    "The product quality is excellent.",
    "I am not happy with this service.",
]

# Passing a LIST processes them in one batch - much faster than a loop.
for sentence, result in zip(sentences, sentiment(sentences)):
    print(f"{result['label']:<9} {result['score']:.3f}   {sentence}")
````

````python
# Now the cases that BREAK sentiment models. Read the confidence scores.

hard = [
    "Oh brilliant, another three-hour lab.",          # sarcasm
    "not bad at all",                                 # double negative
    "The food was ok but the service was terrible.",  # genuinely both
    "It's fine I guess.",                             # lukewarm
]

for sentence, result in zip(hard, sentiment(hard)):
    print(f"{result['label']:<9} {result['score']:.3f}   {sentence}")
````

**✅ Look at the sarcasm line.** The model almost certainly got it wrong — sarcasm depends on tone and context that plain text does not carry.

**And the mixed sentence.** A two-label model is *forced* to pick one. Its confidence score is the only clue it is struggling.

### Example 2 — text generation, and seeing hallucination directly

````python
generator = pipeline("text-generation", model="distilgpt2")

output = generator(
    "Artificial Intelligence is useful because",
    max_new_tokens=40,        # how many NEW tokens to ADD after the prompt
    num_return_sequences=1,
    do_sample=True,           # sample instead of always taking the top token
    temperature=0.8,          # the SAME parameter you met with Gemini
)

print(output[0]["generated_text"])
````

> **`max_new_tokens`, not `max_length`.** `max_length` counts the prompt too, so a long prompt leaves almost nothing for the answer. It is also deprecated. Many older tutorials still use it.

````python
# distilgpt2 is TINY (82M parameters) next to Gemini (billions).
# Run this a few times and read the output critically.
#
# It is grammatical. It is also frequently nonsense.
# That gap between "sounds right" and "is right" IS hallucination -
# and here it is obvious enough to see with your own eyes.

for run in range(3):
    out = generator("The capital of France is", max_new_tokens=20,
                    do_sample=True, temperature=0.9)
    print(f"{run + 1}. {out[0]['generated_text']}\n")
````

### Example 3 — question answering that *cannot* hallucinate

````python
qa = pipeline("question-answering",
              model="distilbert-base-cased-distilled-squad")

context = """
Artificial Intelligence is a branch of computer science that enables machines
to perform tasks that normally require human intelligence. These tasks include
learning, reasoning, problem-solving, understanding language, and recognising
images.
"""

answer = qa(question="What is Artificial Intelligence?", context=context)
print("Answer    :", answer["answer"])
print("Confidence:", round(answer["score"], 3))
print("Found at  : characters", answer["start"], "to", answer["end"])
````

````python
# Now ask something the context does NOT contain.

answer = qa(question="Who invented Artificial Intelligence?", context=context)
print("Answer    :", answer["answer"])
print("Confidence:", round(answer["score"], 4), " <- very low")
````

**✅ This pipeline is *extractive*.** It finds the span of text in the context that best answers the question. It **cannot** produce something the context does not contain.

**That is a feature: it cannot hallucinate.** The low confidence score is how it tells you it is unsure.

> **This is exactly the mechanism behind RAG** — find the relevant text first, then answer only from it. You built the prompt-based version of this in Day 4b.

### Example 4 — the tasks `pipeline()` supports

````python
tasks = pd.DataFrame([
    ("sentiment-analysis",   "Positive or negative?",        "Feedback triage"),
    ("text-classification",  "Which of N categories?",       "Ticket routing"),
    ("question-answering",   "Find the answer in a text",    "Document search"),
    ("summarization",        "Shorten a long text",          "Report digests"),
    ("translation",          "Between languages",            "Localisation"),
    ("ner",                  "Find names, places, dates",    "Data extraction"),
    ("zero-shot-classification", "Categories you invent",    "Flexible tagging"),
    ("text-generation",      "Continue a prompt",            "Drafting"),
    ("fill-mask",            "Guess the hidden word",        "Understanding models"),
    ("image-classification", "What is in this picture?",     "Photo tagging"),
    ("automatic-speech-recognition", "Audio to text",        "Transcription"),
], columns=["pipeline task", "What it does", "A real use"])

tasks
````

````python
# ZERO-SHOT classification is worth knowing: you invent the categories
# at call time, with no training at all.

classifier = pipeline("zero-shot-classification",
                      model="facebook/bart-large-mnli")

result = classifier(
    "The projector kept switching off during the afternoon session.",
    candidate_labels=["equipment problem", "content feedback",
                      "scheduling", "praise"],
)

for label, score in zip(result["labels"], result["scores"]):
    print(f"  {label:<20} {score:.3f}")
````

## 🧠 Quick quiz — The pipeline function

Answer all five from memory. No scrolling back.

**Q1.** Why should you always name the model explicitly in `pipeline()`?

**Q2.** What is the difference between `max_length` and `max_new_tokens`?

**Q3.** Why can the `question-answering` pipeline not hallucinate?

**Q4.** The first run of a pipeline is slow and later runs are instant. Why?

**Q5.** What makes zero-shot classification different from ordinary text classification?

<details>
<summary><b>Answers</b></summary>

**A1.** Without a model name you get whatever default the library currently ships, and that default **changes between versions**. Your results stop being reproducible — a marker re-running your notebook next month may get different numbers.

**A2.** `max_length` counts the prompt **plus** the output, so a long prompt leaves almost nothing for the answer. `max_new_tokens` counts only what is generated. Use `max_new_tokens` — `max_length` is deprecated.

**A3.** It is **extractive**: it finds the span of text *within the context* that best answers the question. It has no ability to produce words that are not in the context. Its low confidence score is how it signals uncertainty.

**A4.** The model is downloaded once (a few hundred MB) and cached in `~/.cache/huggingface`. Later runs load from disk. This is why you install and run once the evening before a session.

**A5.** You invent the categories at call time, with no training. The model was trained to judge whether a sentence entails a statement, so it can score any label you hand it — which is enormously useful when your categories change often.

</details>

# Concept 3 — Choosing a model responsibly

🧠 **Analogy: the store listing.**

Before installing an app you glance at the reviews, the size, the permissions it wants. A **model card** is that listing: what the model does, what it was trained on, how big it is, what licence — and, crucially, **what it is bad at**.

**A model with no card is a model nobody will vouch for.**

> **This is what Responsible AI looks like in practice.** Not a lecture at the end of a course — five minutes reading a page before you use somebody else's model.

## 📘 Examples

### Example 1 — read a card programmatically

````python
from huggingface_hub import model_info

for name in ["distilbert-base-uncased-finetuned-sst-2-english",
             "cardiffnlp/twitter-roberta-base-sentiment-latest"]:
    info = model_info(name)
    print(f"--- {info.id} ---")
    print(f"  downloads (30d): {info.downloads:,}")
    print(f"  likes          : {info.likes}")
    print(f"  tags           : {[t for t in info.tags if '-' not in t][:8]}")
    print()
````

> **Download count is a weak signal, not a strong one.** A popular model is well-tested, but popularity does not make it suitable for *your* text or *your* users. Read the card.

### Example 2 — find the bias the card admits to

````python
# Open this page in a browser and read the "Risks, Limitations and Biases"
# section. Then run this to see the effect for yourself.
#
#   https://huggingface.co/distilbert-base-uncased-finetuned-sst-2-english

probe = [
    "This film was directed by a woman.",
    "This film was directed by a man.",
    "The restaurant serves Mexican food.",
    "The restaurant serves French food.",
]

for sentence, result in zip(probe, sentiment(probe)):
    print(f"{result['label']:<9} {result['score']:.3f}   {sentence}")

print("\nThese sentences are all NEUTRAL statements of fact.")
print("If the model assigns them different sentiment, that difference")
print("came from its training data, not from the sentences.")
````

### 📊 How to read a bias probe

| What you observe | What it means |
|---|---|
| **All four roughly equal** | No obvious bias on this axis — for **these four sentences only** |
| **A consistent gap between paired sentences** | The model learned an association from its training data |
| **High confidence on a neutral sentence** | ⚠️ Worse than a small gap — it is confidently wrong |
| **The gap reverses when you rephrase** | The effect may be phrasing, not the demographic term. Test more |

> ⚠️ **Four sentences is not an audit.** It is a demonstration that the effect exists. A real fairness audit tests hundreds of paired examples and reports metrics per group. **Never claim a model is fair on the basis of a handful of probes** — you can only ever demonstrate that a problem *is* present, never that it is absent.

### Example 3 — the decision checklist

````python
checklist = pd.DataFrame([
    ("Who made it?",             "An unknown author with no card is a risk"),
    ("What was it trained on?",  "Film reviews behave differently from tweets"),
    ("What licence?",            "Some forbid commercial use entirely"),
    ("What limitations are stated?", "If the card admits bias, believe it"),
    ("How many downloads?",      "Weak signal - popular is not the same as suitable"),
    ("When was it last updated?","An abandoned model will not be fixed"),
    ("Does it match MY text?",   "The single most important question"),
], columns=["Question", "Why it matters"])

checklist
````

### The three decisions to make every time

Ask these before using any model, and write the answers down:

| Could I use this model for… | Think about |
|---|---|
| **Medical triage?** | Almost never. A stated bias plus a health decision is unacceptable |
| **Screening job applications?** | Almost never. Bias here reproduces discrimination at scale |
| **A commercial product?** | Check the **licence** first, then the limitations |

**For the sentiment model above, the card explicitly warns about biased predictions on certain demographic and country terms.** That one line changes the answers: fine for a classroom demo or ranking film reviews; **not fine** for anything affecting someone's healthcare or employment.

## 🧠 Quick quiz — Choosing a model responsibly

Answer all five from memory. No scrolling back.

**Q1.** What is a model card, and what is the most important section?

**Q2.** Why is 'it has 10 million downloads' a weak reason to trust a model?

**Q3.** A model card admits the model shows bias on certain demographic terms. Can you use it to rank film reviews?

**Q4.** You probe a model with four sentences and see no bias. Is the model fair?

**Q5.** Which is more dangerous: a model that is wrong with low confidence, or wrong with high confidence?

<details>
<summary><b>Answers</b></summary>

**A1.** The documentation page for a model: who made it, training data, licence, intended use, and limitations. **The limitations section is the most important** — it is the part that tells you where the model fails, which is the part that will hurt you.

**A2.** Popularity means it is well-tested for the tasks *other people* use it for. It says nothing about whether it suits **your** text, your users, or your risk tolerance. A model trained on film reviews is popular and still wrong for medical notes.

**A3.** Probably yes — that is close to what it was trained for, and the stakes are low. But you would still note the limitation, and you would **not** use it for anything affecting a person's healthcare, employment or credit.

**A4.** **No, and you must not claim it is.** Four sentences is a demonstration, not an audit. You can show a problem is present; you can never show it is absent from a handful of examples. Real audits test hundreds of paired cases and report metrics per group.

**A5.** **High confidence.** A low-confidence wrong answer can be caught by a threshold and routed to a human. A confidently wrong answer looks exactly like a confidently right one, so nothing downstream flags it.

</details>

Find a model on the Hub that has not been updated in over two years.

1. Does it still work with current library versions?
2. Does its card mention which versions it was tested against?
3. What would you do if it broke in production and nobody maintains it?
4. Write a short policy for your capstone team: **what must be true before you build on someone else's model?**
