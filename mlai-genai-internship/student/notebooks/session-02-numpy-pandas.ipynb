# 1. NumPy fundamentals

🧠 **The shopping bag and the egg tray.** A list is a bag — to act on everything you pull items out one at a time. An array is an egg tray — every slot is the same kind of thing, so the computer acts on **all of them at once**.

```python
prices_list  = [100, 200, 300]
prices_array = np.array([100, 200, 300])

print("list  * 2 =", prices_list * 2)
print("            ^ the BAG was duplicated")
print()
print("array * 2 =", prices_array * 2)
print("            ^ every ELEMENT was doubled  <- what we want")
```

```python
a = np.array([1, 2, 3, 4, 5])
b = np.array([[1, 2, 3],
              [4, 5, 6]])

print("a.shape :", a.shape, "  5 elements, 1 dimension")
print("b.shape :", b.shape, " 2 rows, 3 columns")
print("a.dtype :", a.dtype)

print("\na + 10 :", a + 10)      # no loop anywhere
print("a ** 2 :", a ** 2)
print("\nb transposed:\n", b.T)
print("\nreshape:\n", np.arange(12).reshape(3, 4))
```

```python
data = np.array([23, 45, 12, 67, 34, 89, 21])

print("mean  :", data.mean().round(2))
print("median:", np.median(data))
print("std   :", data.std().round(2))
print("min/max:", data.min(), data.max())

# BOOLEAN MASKING - the idea behind every Pandas filter you will write
mask = data > 30
print("\ndata > 30  :", mask)
print("data[mask] :", data[mask], " only the True positions")
print("how many   :", mask.sum(), " (True counts as 1)")
```

```python
# Reproducible randomness: fix the seed so results can be compared.
rng = np.random.default_rng(seed=42)
print("seed 42:", rng.integers(1, 100, size=5))

rng2 = np.random.default_rng(seed=42)
print("again  :", rng2.integers(1, 100, size=5), " <- identical")
```

# 2. Pandas for data handling

🧠 **A spreadsheet that takes orders.** Everything you can do in Excel has a one-line Pandas equivalent.

```python
df = pd.read_csv(BASE + "regression/cardekho_dataset.csv")
print("Loaded", df.shape[0], "cars,", df.shape[1], "columns")
df.head(3)
```

```python
# ONE name in [] gives a Series. A LIST of names gives a DataFrame.
print("one column  ->", type(df["selling_price"]).__name__)
print("two columns ->", type(df[["brand", "selling_price"]]).__name__)

df[["car_name", "brand", "selling_price", "fuel_type"]].head()
```

```python
# FILTERING: use & for AND, | for OR, and wrap EACH condition in brackets.
# Plain `and` / `or` do not work on whole columns.

expensive = df[df["selling_price"] > 1000000]
recent_petrol = df[(df["vehicle_age"] < 5) & (df["fuel_type"] == "Petrol")]

print("cars above 10 lakh :", len(expensive))
print("recent petrol cars :", len(recent_petrol))
recent_petrol[["car_name", "vehicle_age", "selling_price"]].head()
```

```python
# GROUPING is the single most useful Pandas skill.
print("Average price by fuel type:")
print(df.groupby("fuel_type")["selling_price"].mean().round(0).sort_values(ascending=False))

print("\nSeveral questions at once with agg():")
print(df.groupby("brand")["selling_price"].agg(["count", "mean", "max"]).head(5).round(0))
```

```python
# A new column from existing ones
df["price_lakhs"] = (df["selling_price"] / 100000).round(2)

print("Three most expensive:")
print(df.sort_values("selling_price", ascending=False)[["car_name", "price_lakhs"]].head(3))

print("\nHow many of each fuel type:")
print(df["fuel_type"].value_counts())
print("\nAs proportions:")
print(df["fuel_type"].value_counts(normalize=True).round(3))
```

**Five commands, on every single new dataset, in this order.**

```python
# 5. What is missing, and where?
missing = loans.isnull().sum()
print(missing[missing > 0] if missing.sum() else "No missing values.")
```

### Twelve rows you can check by eye

Before working with 10,000 rows, work with one you can hold in your head.

```python
small = pd.read_csv(BASE + "prepreprocessing/pre_data.csv")
small
```

```python
print("--- missing values per column ---")
print(small.isnull().sum())

print("\n--- duplicate rows ---")
print(small[small.duplicated(keep=False)])

print("\n--- numeric summary: compare 75% against max ---")
print(small.describe().round(1))
```

**✅ Two things to notice in that last table.**

1. Salary `75%` is **81,000** but `max` is **1,500,000**. *A max many times the 75th percentile is the signature of an outlier.*
2. Salary `mean` is about **196,091** but the median (`50%`) is **67,000**. One row moved the average by nearly 130,000.

**That second point is the concrete reason we fill missing numbers with the median, not the mean.**

# 4. Exploratory Data Analysis

🧠 **The doctor's check-up.** A doctor does not prescribe the moment you walk in. Skipping EDA and jumping to `model.fit()` is prescribing without examining the patient.

```python
# Is the target balanced? This decides which metrics you can trust.
print(loans["loan_status"].value_counts())
print()
print(loans["loan_status"].value_counts(normalize=True).round(3))
```

```python
# Which numeric columns relate to the target?
#
# Near +1 or -1 -> moves WITH the target
# Near 0        -> no LINEAR relationship ON ITS OWN
#
# NOTE: correlation is NOT causation. Ice cream sales correlate with
# drowning deaths - hot weather causes both.

loans.corr(numeric_only=True)["loan_status"].sort_values().round(3)
```

```python
# A group comparison often says more than a correlation number.
print("Average credit score by outcome:")
print(loans.groupby("loan_status")["credit_score"].mean().round(1))

print("\nAverage loan-to-income by outcome:")
loans["loan_to_income"] = loans["loan_amnt"] / loans["person_income"]
print(loans.groupby("loan_status")["loan_to_income"].mean().round(4))
```
