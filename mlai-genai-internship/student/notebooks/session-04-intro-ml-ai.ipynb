# 1. What Machine Learning actually is

🧠 **The recipe and the chef.** A recipe is exact instructions you follow. A chef tastes a thousand curries and works out the rules for themselves. **Nobody wrote those rules down.**

```text
Traditional programming:   Data + RULES    ->  Answers
Machine Learning:          Data + ANSWERS  ->  Rules
```

````python
# The rule you cannot write.
def is_spam(text):
    if "lottery" in text.lower():
        return True
    if "free money" in text.lower():
        return True
    return False

print(is_spam("You have WON a lottery prize!"))    # True
print(is_spam("Cheap m3dicine, no prescription"))  # False -- and it IS spam

print("\nThe spammer wrote m3dicine to dodge the keyword.")
print("You cannot write a rule for every spelling.")
````

````python
# The same problem, LEARNED from examples instead.
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

emails = ["win a free prize now", "claim your lottery money",
          "meeting moved to 3pm", "notes from the lecture attached",
          "cheap m3dicine no prescription", "fee payment reminder"]
labels = [1, 1, 0, 0, 1, 0]        # 1 = spam

vec = CountVectorizer()
model = MultinomialNB().fit(vec.fit_transform(emails), labels)

tests = ["free money claim now", "lecture notes for tomorrow"]
print("predictions:", model.predict(vec.transform(tests)), " (1 = spam)")
print("\nNobody wrote a rule. Six examples were enough.")
````

````python
# What "learning" looks like as a number.
from sklearn.linear_model import LinearRegression

# Hours studied -> exam mark. The true relationship is mark = 8*hours + 30
hours = np.array([[1], [2], [3], [4], [5], [6]])
marks = np.array([38, 46, 54, 62, 70, 78])

m = LinearRegression().fit(hours, marks)
print(f"learned: mark = {m.coef_[0]:.1f} * hours + {m.intercept_:.1f}")
print("\nThe model was never told the formula. It recovered it from six examples.")
````

# 3. Types of Machine Learning

| Type | You give it | It learns to |
|---|---|---|
| **Supervised** | Inputs **and** correct answers | Predict the answer for new inputs |
| **Unsupervised** | Inputs only, no answers | Find structure and groups |
| **Reinforcement** | An environment and a reward | Act to maximise reward |
| **Generative** | Huge amounts of text/images | Produce new content |

**Supervised splits in two:** *regression* if the answer is a **number**, *classification* if it is a **category**.

````python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.cluster import KMeans

loans = pd.read_csv(BASE + "loan_data_10k.csv").dropna().reset_index(drop=True)
for c in loans.select_dtypes(include="object").columns:
    loans[c] = LabelEncoder().fit_transform(loans[c])

X = loans.drop(columns=["loan_status"])
y = loans["loan_status"]
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=.2, random_state=42, stratify=y)

# SUPERVISED - we hand it the answers (y)
clf = LogisticRegression(max_iter=2000).fit(Xtr, ytr)
print("supervised classification:", round(clf.score(Xte, yte), 4))

# UNSUPERVISED - notice there is NO y. No answers to learn from.
groups = KMeans(n_clusters=3, n_init=10, random_state=42).fit_predict(X)
print("unsupervised groups found:", np.unique(groups), "-> numbers it invented")
````

````python
questions = [
    ("How much will this house sell for?",          "Supervised - Regression"),
    ("Will this customer churn?",                   "Supervised - Classification"),
    ("What natural groups exist in our customers?", "Unsupervised - Clustering"),
    ("Write a product description",                 "Generative"),
    ("How many visitors tomorrow?",                 "Supervised - Regression"),
    ("Which of 5 plans will they choose?",          "Supervised - Classification"),
]
for q, a in questions:
    print(f"{q:<46} {a}")

print("\nDecide NUMBER or CATEGORY and you have made the single most")
print("important modelling decision - it fixes your models AND your metrics.")
````

# 4. The Machine Learning workflow

```text
1. Define the problem   -> Number or category?
2. Collect data         -> Where from? Am I allowed to use it?
3. Explore (EDA)        -> shape, info, missing, target balance
4. Preprocess           -> clean, encode, split, scale     <- most of your time
5. Choose a model       -> start with the simplest that could work
6. Train                -> model.fit(X_train, y_train)
7. Evaluate             -> on data the model has NEVER seen
8. Improve              -> features, cross-validation, tuning
9. Deploy               -> a Streamlit app, an API
10. Monitor             -> does it still work on next month's data?
```

**Roughly 70% of real ML work is steps 2 to 4.** Nobody puts that in the demo video.

````python
# The four methods that never change.
#
#   model = SomeModel(...)        1. create, with settings
#   model.fit(X_train, y_train)   2. learn from training data
#   model.predict(X_test)         3. predict on new data
#   model.score(X_test, y_test)   4. a quick default score
#
# Learn these once and EVERY scikit-learn model works the same way.

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report

model = RandomForestClassifier(n_estimators=200, random_state=42)
model.fit(Xtr, ytr)
print(classification_report(yte, model.predict(Xte)))
````

````python
# ALWAYS compute a baseline first. It costs three lines.
from sklearn.dummy import DummyClassifier

baseline = DummyClassifier(strategy="most_frequent").fit(Xtr, ytr)

b = baseline.score(Xte, yte)
m = model.score(Xte, yte)
print(f"baseline (always guess majority) : {b:.4f}")
print(f"random forest                    : {m:.4f}")
print(f"gain                             : {m - b:+.4f}")
print("\nIf your model cannot clearly beat the baseline, it has learned nothing.")
````

# 5. ML & AI APIs

🧠 **Choosing a vehicle.** A bicycle for the corner shop, a van for moving house, a truck for freight. All three "move things" — and picking the truck for a loaf of bread is a mistake.

| Library | Built for | Use it when |
|---|---|---|
| **scikit-learn** | Classical ML on **tables** | Rows and columns, under ~1M rows |
| **TensorFlow** | Deep learning at production scale | Large deployments, mobile |
| **Keras** | A friendly **front end** for TensorFlow | Building networks quickly |
| **PyTorch** | Deep learning, research-first | Most new DL work; full control |

> **Keras is not a competitor to TensorFlow — it runs on top of it.** You write Keras; TensorFlow does the work underneath.

````python
# The same task in scikit-learn: three lines.
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import StandardScaler
import time

sc = StandardScaler()
Xtr_s, Xte_s = sc.fit_transform(Xtr), sc.transform(Xte)

t0 = time.time()
mlp = MLPClassifier(hidden_layer_sizes=(16, 8), max_iter=500, random_state=42)
mlp.fit(Xtr_s, ytr)
mlp_time = time.time() - t0

t0 = time.time()
rf = RandomForestClassifier(n_estimators=200, random_state=42).fit(Xtr, ytr)
rf_time = time.time() - t0

print(f"{'model':<16}{'score':>8}{'seconds':>10}")
print("-" * 34)
print(f"{'MLPClassifier':<16}{mlp.score(Xte_s, yte):>8.4f}{mlp_time:>10.1f}")
print(f"{'RandomForest':<16}{rf.score(Xte, yte):>8.4f}{rf_time:>10.1f}")
print("\nOn TABULAR data the forest usually wins - and trains far faster.")
print("Do not reach for deep learning because it sounds more advanced.")
````
