```python
books = {
    "B001": {"title": "Python Basics",    "author": "Smith", "available": True,  "borrower": None},
    "B002": {"title": "Machine Learning", "author": "Kumar", "available": False, "borrower": "Ravi"},
    "B003": {"title": "Data Science",     "author": "Nair",  "available": True,  "borrower": None},
}

def list_books(books):
    for book_id, b in books.items():
        state = "available" if b["available"] else f"with {b['borrower']}"
        print(f"{book_id}  {b['title']:<20} {b['author']:<8} {state}")

def find_by_author(books, author):
    return [b["title"] for b in books.values() if b["author"] == author]

def borrow(books, book_id, borrower):
    if book_id not in books:
        return f"No book with id {book_id}"
    book = books[book_id]
    if not book["available"]:
        return f"'{book['title']}' is already with {book['borrower']}"
    book["available"] = False
    book["borrower"] = borrower
    return f"{borrower} borrowed '{book['title']}'"

def return_book(books, book_id):
    if book_id not in books:
        return f"No book with id {book_id}"
    book = books[book_id]
    if book["available"]:
        return f"'{book['title']}' was not borrowed"
    who = book["borrower"]
    book["available"] = True
    book["borrower"] = None
    return f"{who} returned '{book['title']}'"

def available_count(books):
    return sum(1 for b in books.values() if b["available"])

list_books(books)
print("\nBy Smith:", find_by_author(books, "Smith"))
print("By Unknown:", find_by_author(books, "Unknown"))
print()
print(borrow(books, "B001", "Asha"))
print(borrow(books, "B002", "Meera"))     # already borrowed
print(borrow(books, "B999", "Sam"))       # does not exist
print(return_book(books, "B001"))
print(return_book(books, "B003"))         # was not borrowed
print("\nAvailable:", available_count(books))
```

```python
days  = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temps = [31.2, 28.5, 35.0, 29.8, 33.4, 27.1, 36.7]

def c_to_f(c):
    return c * 9 / 5 + 32

def f_to_c(f):
    return (f - 32) * 5 / 9

def describe(c):
    if c > 35:
        return "Very Hot"
    if c >= 30:
        return "Hot"
    if c >= 20:
        return "Pleasant"
    return "Cold"

def hottest_day(days, temps):
    return days[temps.index(max(temps))]

def coldest_day(days, temps):
    return days[temps.index(min(temps))]

def weekly_report(days, temps):
    avg = sum(temps) / len(temps)
    hot_days = sum(1 for t in temps if t >= 30)
    print(f"average      : {avg:.1f} C")
    print(f"hottest      : {hottest_day(days, temps)} ({max(temps)} C)")
    print(f"coldest      : {coldest_day(days, temps)} ({min(temps)} C)")
    print(f"hot or worse : {hot_days} days")

print(c_to_f(0), c_to_f(100))
print(f_to_c(c_to_f(25)))

print(f"\n{'DAY':<5}{'C':>7}{'F':>8}  DESCRIPTION")
print("-" * 36)
for day, c in zip(days, temps):
    print(f"{day:<5}{c:>7.1f}{c_to_f(c):>8.1f}  {describe(c)}")

print()
weekly_report(days, temps)
```

```python
raw = [
    {"name": "Asha",  "age": 21,   "city": "Delhi",   "score": 85},
    {"name": "Ravi",  "age": 22,   "city": "mumbai",  "score": None},
    {"name": "Meera", "age": None, "city": "DELHI",   "score": 92},
    {"name": "John",  "age": 23,   "city": "Chennai", "score": 61},
    {"name": "Farah", "age": 21,   "city": " delhi ", "score": 78},
    {"name": "John",  "age": 23,   "city": "Chennai", "score": 61},
]

# Task 2
print("rows          :", len(raw))
print("missing age   :", sum(1 for r in raw if r["age"] is None))
print("missing score :", sum(1 for r in raw if r["score"] is None))

seen, duplicates = [], 0
for r in raw:
    key = tuple(r.items())
    if key in seen:
        duplicates += 1
    else:
        seen.append(key)
print("duplicates    :", duplicates)

# Task 3
clean, seen = [], set()
for r in raw:
    key = tuple(r.items())
    if key not in seen:
        seen.add(key)
        clean.append(dict(r))
print(f"\nRemoved {len(raw) - len(clean)} duplicate row(s) -> {len(clean)} rows")

# Task 4
for r in clean:
    r["city"] = r["city"].strip().title()
print("distinct cities:", sorted({r["city"] for r in clean}))

# Task 5
def median(values):
    values = sorted(v for v in values if v is not None)
    mid = len(values) // 2
    if len(values) % 2:
        return values[mid]
    return (values[mid - 1] + values[mid]) / 2

median_age = median([r["age"] for r in clean])
median_score = median([r["score"] for r in clean])
print(f"\nfilling age with {median_age}, score with {median_score}")

for r in clean:
    if r["age"] is None:
        r["age"] = median_age
    if r["score"] is None:
        r["score"] = median_score

# Task 6
cities = sorted({r["city"] for r in clean})
mapping = {city: i for i, city in enumerate(cities)}
print("city mapping:", mapping)
for r in clean:
    r["city_code"] = mapping[r["city"]]

# Task 7
train, test = clean[:4], clean[4:]
print(f"\ntrain {len(train)} rows | test {len(test)} rows")
print("\nFinal cleaned data:")
for r in clean:
```
