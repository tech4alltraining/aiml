# 1. Variables and printing

🧠 **A variable is a labelled box.** `=` means *put this value in this box* — an instruction, not a statement of fact.

````python
# Creating variables. Python works out the type for you.
name = "Asha"
age = 21
height = 5.4
is_student = True

print(name, age, height, is_student)

# A box keeps only the LAST value you put in it.
score = 50
score = 87
print("score is", score)     # 87 - the 50 is gone
````

````python
# f-strings: put an f before the quote, then {variable} inside.
name, marks = "Asha", 87.456

print("Student", name, "scored", marks)        # clumsy, with commas
print(f"Student {name} scored {marks}")        # f-string
print(f"Student {name} scored {marks:.1f}")    # :.1f = one decimal place
print(f"{name} needs {100 - marks:.1f} more")  # maths inside the braces
````

````python
# Alignment makes tabular output readable.
#   :<10  left-align in 10 characters      :>7  right-align in 7

students = {"Asha": 87.4, "Ravi": 74.15, "Meera": 95.5}

print(f"{'NAME':<10}{'MARK':>7}")
print("-" * 17)
for name, mark in students.items():
    print(f"{name:<10}{mark:>7.1f}")
````

# 2. Data types

| Type | Example | What it is |
|---|---|---|
| `str` | `"Asha"` | Text. **Always in quotes** |
| `int` | `21` | A whole number |
| `float` | `5.4` | A decimal |
| `bool` | `True` | Yes or no. **Capital** T and F |

````python
for value in ["Asha", 21, 5.4, True]:
    print(f"{str(value):<8} {type(value).__name__}")
````

````python
# THE most common beginner error: text that looks like a number.
print("21" + "5")     # "215"  - text is GLUED together
print(21 + 5)         # 26     - numbers are ADDED

try:
    print("21" + 5)   # Python will NOT guess
except TypeError as error:
    print("ERROR:", error)
````

````python
# Converting between types
print(int("21") + 5)        # 26
print(float("3.14") * 2)    # 6.28
print("Age: " + str(21))    # "Age: 21"

print(int(7.9))             # 7  <- CHOPS the decimal, does NOT round
print(round(7.9))           # 8  <- use round() if you want rounding

# int("12.5") fails: not a whole number in text form. Convert in two steps.
print(int(float("12.5")))   # 12
````

> ⚠️ **`input()` always returns a string**, even when the user typed digits. Convert it: `age = int(input(...))`. This is the single most common beginner error in Python.

# 3. Operators

⚠️ **`=` assigns. `==` compares.** Mixing them up is the most common typo in programming.

````python
a, b = 17, 5

print(f"{a} + {b}  = {a + b}")
print(f"{a} / {b}  = {a / b}      <- normal division ALWAYS gives a float")
print(f"{a} // {b} = {a // b}          <- floor division: the whole part")
print(f"{a} % {b}  = {a % b}          <- modulus: the REMAINDER")
print(f"{a} ** 2 = {a ** 2}")

print()
print("Why % matters - it tells you if a number divides evenly:")
print(f"  10 % 2 = {10 % 2}  -> 10 is EVEN")
print(f"  11 % 2 = {11 % 2}  -> 11 is ODD")
````

````python
marks, attendance = 78, 65

# Comparison gives back True or False
print("marks == 78 ->", marks == 78)
print("marks >= 75 ->", marks >= 75)

# Logical operators combine conditions
print("passed AND good attendance:", marks >= 40 and attendance >= 75)
print("passed OR  good attendance:", marks >= 40 or attendance >= 75)
print("did NOT pass             :", not (marks >= 40))

# Membership works on lists AND on strings
subjects = ["Maths", "Physics", "Python"]
print("\n'Python' in subjects :", "Python" in subjects)
print("'y' in 'Python'      :", "y" in "Python")
````

# 4. Conditionals and loops

⚠️ **Indentation is not decoration in Python — it is the syntax.**

````python
# WHY elif AND NOT A SECOND if
m = 95

g = ""
if m >= 90:
    g = "A"
if m >= 75:        # this ALSO runs, and overwrites the A
    g = "B"
print("two ifs  :", g, " <- WRONG")

g = ""
if m >= 90:
    g = "A"
elif m >= 75:      # only runs if the first was False
    g = "B"
print("with elif:", g, " <- correct")
````

````python
marks = [78, 92, 65, 88]
subjects = ["Maths", "Physics", "Python", "Stats"]

print("--- values ---")
for mark in marks:
    print(mark)

print("\n--- enumerate: position AND value ---")
for i, mark in enumerate(marks):
    print(f"Subject {i + 1}: {mark}")

print("\n--- zip: two lists together ---")
for sub, mark in zip(subjects, marks):
    status = "pass" if mark >= 75 else "needs work"
    print(f"{sub:<10}{mark:>4}  {status}")

print("\n--- range ---")
print(list(range(1, 6)), list(range(0, 10, 2)))
````

````python
# Building a result: start empty, add as you go.
passed = []
for mark in marks:
    if mark >= 60:
        passed.append(mark)
print("long way  :", passed)

# A list comprehension does the same in one line.
print("short way :", [m for m in marks if m >= 60])
print("with maths:", [m + 5 for m in marks])

# while: repeat until a condition stops being true
countdown = 3
while countdown > 0:
    print("countdown:", countdown)
    countdown -= 1          # MUST change, or it loops forever
print("go!")
````

# 5. Collections

| Type | Analogy | Ordered? | Changeable? | Duplicates? |
|---|---|---|---|---|
| **list** `[ ]` | A shelf of books | Yes | Yes | Yes |
| **tuple** `( )` | A sealed box of coordinates | Yes | **No** | Yes |
| **dict** `{k:v}` | A labelled drawer | Yes | Yes | Keys unique |
| **set** `{ }` | A bag of raffle tickets | No | Yes | **No** |

````python
marks = [78, 92, 65, 88, 45]

print("the list   :", marks)
print("first/last :", marks[0], marks[-1])
print("slice 1:3  :", marks[1:3], " includes 1, EXCLUDES 3")

marks.append(97)
marks[0] = 80
marks.remove(45)
print("after edits:", marks)

print(f"\nsum {sum(marks)} | max {max(marks)} | min {min(marks)} "
      f"| average {sum(marks) / len(marks):.1f}")
````

````python
student = {"name": "Asha", "age": 21, "cgpa": 8.45}

print(student["name"])
print("missing key with .get():", student.get("phone"))
print("with a default          :", student.get("phone", "not provided"))

student["semester"] = 6
student["cgpa"] = 8.60

print()
for key, value in student.items():
    print(f"  {key:<10}: {value}")
````

````python
# SET: duplicates vanish. Perfect for "what distinct values exist?"
cities = ["Delhi", "Mumbai", "Delhi", "Chennai", "Mumbai", "Delhi"]
print("original :", cities)
print("as a set :", set(cities))
print("distinct :", len(set(cities)))

# Sets do maths on groups
python_students = {"Asha", "Ravi", "Meera"}
ml_students = {"Ravi", "Meera", "John"}
print("\nin BOTH    :", python_students & ml_students)
print("in EITHER  :", python_students | ml_students)
print("Python only:", python_students - ml_students)

# TUPLE: fixed. Use when values must not change.
location = (10.85, 76.27)
try:
    location[0] = 99
except TypeError as error:
    print("\nCannot change a tuple:", error)
````

# 6. Strings

🧠 **A row of numbered letter tiles.** Each character has a position, starting at **0**.

````python
text = "Machine Learning"
#       0123456789...

print("length    :", len(text))
print("text[0]   :", text[0],    "   first character")
print("text[-1]  :", text[-1],   "   last character")
print("text[0:7] :", text[0:7],  " includes 0, EXCLUDES 7")
print("text[8:]  :", text[8:],   "  from 8 to the end")
````

````python
messy = "  Machine Learning IS Fun!  "

print(f"original  : {messy!r}")
print(f".strip()  : {messy.strip()!r}")
print(f".lower()  : {messy.strip().lower()!r}")
print(f".replace(): {messy.strip().replace('Fun', 'Useful')!r}")
print(f".split()  : {messy.split()}")
print(f"count 'n' : {messy.lower().count('n')}")
````

````python
# THE IMMUTABILITY TRAP - strings never change in place.
t = "hello"
t.upper()          # returns "HELLO" ... and throws it away
print("not captured:", t)

t = t.upper()      # you must CAPTURE the result
print("captured   :", t)
````

````python
# The real ML use of strings: making inconsistent categories match.
raw = ["  Delhi", "delhi ", "DELHI", "Delhi"]
cleaned = [city.strip().lower() for city in raw]

print("raw     :", raw)
print("cleaned :", cleaned)
print(f"\n{len(set(raw))} distinct values -> {len(set(cleaned))} after cleaning")

# Splitting a CSV line - you will do this constantly.
row = "Asha,21,Computer Science,8.45"
name, age, course, cgpa = row.split(",")
print(f"\n{name} is {age} and studies {course}")
````

# 7. Functions

🧠 **A recipe card.** Write it once, name it, reuse it.

**`print` shows a value. `return` hands it back** so the caller can keep using it.

````python
def greet(name):
    """Print a greeting. This text is the docstring."""
    print(f"Hello, {name}!")

greet("Asha")      # `def` only DEFINES. Calling is what runs it.

def calculate_percentage(obtained, total=100):
    """Return the percentage, rounded to 2 places."""
    return round((obtained / total) * 100, 2)

print(calculate_percentage(87))              # uses the default total
print(calculate_percentage(87, total=120))   # overrides it

# Because it RETURNS, you can keep using the value:
print("doubled:", calculate_percentage(45) * 2)
````

````python
def student_report(name, marks, passing=40, show_grade=True):
    """Build a one-line report."""
    status = "PASS" if marks >= passing else "FAIL"
    line = f"{name:<10} {marks:>3}  {status}"
    if show_grade:
        grade = "A" if marks >= 90 else "B" if marks >= 75 else "C"
        line += f"  grade {grade}"
    return line

print(student_report("Asha", 87))                     # positional
print(student_report(name="Ravi", marks=35))          # keyword - clearer
print(student_report("Meera", 92, show_grade=False))
````

````python
# Small functions that build on each other. Each does ONE job.

def average(numbers):
    """Return the average, or 0 for an empty list."""
    if not numbers:
        return 0
    return sum(numbers) / len(numbers)

def grade_from(mark):
    if mark >= 90: return "A"
    if mark >= 75: return "B"
    if mark >= 60: return "C"
    return "F"

def class_report(students):
    all_averages = []
    for name, marks in students.items():
        avg = average(marks)
        all_averages.append(avg)
        print(f"  {name:<10} {avg:>6.1f}  {grade_from(avg)}")
    print(f"  {'CLASS':<10} {average(all_averages):>6.1f}")

class_report({"Asha": [78, 92, 65], "Ravi": [88, 74, 91], "Meera": [95, 89, 97]})
````

# 8. Classes and objects

🧠 **Analogy: the blueprint and the houses.**

A **class** is a blueprint. It is not a thing you can live in — it describes what every house of that kind will have.

An **object** is a house built from that blueprint. Build a hundred houses from one blueprint and each has its **own** rooms with its own furniture.

```text
class Student:          <- the blueprint (written once)
      |
      +-- asha  = Student("Asha", 21)     <- an object
      +-- ravi  = Student("Ravi", 22)     <- another object
```

**Why bother?** Once you have three or four related things — a name, marks, a method that grades them — passing them around separately gets messy. A class keeps **the data and the things you do to it** in one place.

> You meet this constantly in ML without writing it yourself. `LinearRegression()` is a class, `model = LinearRegression()` builds an object, `model.fit(...)` calls its method.

````python
class Student:
    """A student with a name, an age, and some marks."""

    def __init__(self, name, age):
        # __init__ runs automatically when you build an object.
        # `self` is the object being built - "this particular student".
        self.name = name        # ATTRIBUTES: data belonging to THIS object
        self.age = age
        self.marks = []

    def add_mark(self, mark):
        """A METHOD: a function that belongs to the class."""
        self.marks.append(mark)

    def average(self):
        if not self.marks:
            return 0
        return sum(self.marks) / len(self.marks)


asha = Student("Asha", 21)     # build an object from the blueprint
ravi = Student("Ravi", 22)     # a second, completely SEPARATE object

asha.add_mark(78)
asha.add_mark(92)
ravi.add_mark(55)

print(f"{asha.name}: marks {asha.marks}, average {asha.average()}")
print(f"{ravi.name}: marks {ravi.marks}, average {ravi.average()}")
print("\nEach object has its OWN marks list - they do not share.")
````

````python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def __str__(self):
        """What print() should show for this object."""
        return f"'{self.title}' by {self.author} ({self.pages}p)"

    def is_long(self):
        return self.pages > 300


b = Book("Python Basics", "Smith", 420)
print(b)              # readable, thanks to __str__
print("Long book?", b.is_long())

# Without __str__ you would get something like:
#   <__main__.Book object at 0x10f3c2d50>  -- which tells a reader nothing
````

````python
# The pattern you will meet in EVERY scikit-learn model.

class SimpleAverager:
    """A deliberately tiny model, to show the shape of the sklearn API."""

    def __init__(self):
        self.prediction = None          # nothing learned yet

    def fit(self, y):
        """Learn from data - here, just remember the average."""
        self.prediction = sum(y) / len(y)
        return self                     # sklearn returns self so you can chain

    def predict(self, n):
        """Use what was learned."""
        if self.prediction is None:
            raise ValueError("Call fit() before predict()")
        return [self.prediction] * n


model = SimpleAverager()        # 1. create
model.fit([10, 20, 30, 40])     # 2. learn from data
print(model.predict(3))         # 3. predict

print()
print("This is EXACTLY the shape of every scikit-learn model:")
print("  model = LinearRegression()")
print("  model.fit(X_train, y_train)")
print("  model.predict(X_test)")
````

````python
# Calling predict() before fit() raises a clear error - by design.
fresh = SimpleAverager()
try:
    fresh.predict(3)
except ValueError as error:
    print("ValueError:", error)
````
