# 4. Data types

🧠 **The difference between a number and a phone number.** Both are made of digits. You can add two numbers meaningfully; adding two phone numbers is nonsense. **The type tells Python which operations make sense.**

| Type | Example | For |
|---|---|---|
| **int** | `42` | Whole numbers |
| **float** | `3.14` | Numbers with decimals |
| **str** | `"hello"` | Text |
| **bool** | `True` | Yes/no |

````python
count = 10               # int
price = 99.50            # float
name = "Kochi"           # str
is_open = True           # bool

print(type(count))
print(type(price))
print(type(name))
print(type(is_open))
````

````python
# Text and number are genuinely different things.
print("5" + "3")         # 53   - joins two pieces of TEXT
print(5 + 3)             # 8    - adds two NUMBERS
print()
print("The + symbol means JOIN for text and ADD for numbers.")
print("The TYPE is what decides which one happens.")
````

````python
# Converting between types.
print(int("42"))         # 42     text -> whole number
print(float("3.14"))     # 3.14   text -> decimal
print(str(42))           # "42"   number -> text
print(int(3.9))          # 3      CHOPS the decimal - it does NOT round
print(round(3.9))        # 4      this is how you round
print(bool(0))           # False  0 is False, every other number is True
````

# 5. Numbers

| Operator | Meaning | `7 op 3` |
|---|---|---|
| `+` `-` `*` | Add, subtract, multiply | `10`, `4`, `21` |
| `/` | Divide | `2.333...` |
| `//` | Floor divide | `2` |
| `%` | Remainder | `1` |

> ⚠️ **`/` always gives a float**, even when it divides evenly. Use `//` for a whole number.

````python
# % is how you test for even numbers. You will use this constantly.
print(8 % 2)      # 0  -> EVEN numbers leave nothing over
print(9 % 2)      # 1  -> ODD numbers leave 1
print()
print("Order of operations:")
print(2 + 3 * 4)        # 14 - multiplication happens first
print((2 + 3) * 4)      # 20 - brackets force the addition first
````

# 7. Strings

🧠 **A row of numbered boxes, one letter per box.** You can look inside any box or take a run of boxes. **What you cannot do is change a box — Python strings never change.**

```text
text = "PYTHON"
        012345      <- positions from the left
       -654321      <- positions from the right
```

````python
text = "PYTHON"

print(text[0])      # P   - the FIRST character (positions start at 0)
print(text[3])      # H
print(text[-1])     # N   - the LAST character
print(text[-2])     # O
print(len(text))    # 6
````

````python
# Slicing: text[start:stop] - START is included, STOP is NOT.
text = "PYTHON"
print(text[0:3])    # PYT   - positions 0, 1, 2  (NOT 3)
print(text[2:5])    # THO
print(text[:3])     # PYT   - from the beginning
print(text[3:])     # HON   - to the end
print(text[-3:])    # HON   - the last three
print()
print("The stop is EXCLUDED. That is why [0:3] and [3:] fit together")
print("perfectly, with no overlap and nothing missed.")
````

````python
# Strings are IMMUTABLE - every method returns a NEW string.
name = "priya sharma"

print(name.upper())        # PRIYA SHARMA
print(name.title())        # Priya Sharma
print(name)                # priya sharma  <- the ORIGINAL is unchanged

name = name.title()        # to KEEP the result, you must assign it
print(name)                # Priya Sharma
````

````python
# The methods you will actually use.
messy = "   Priya Sharma   "
print(f"[{messy.strip()}]")                    # remove surrounding spaces
print("a-b-c".replace("-", " "))               # swap text
print("a,b,c".split(","))                      # break into a list
print("hello".find("l"))                       # 2  - position, or -1
print("hello".count("l"))                      # 2  - how many times
print("report.csv".endswith(".csv"))           # True
````

````python
# Escape characters.
print("Line one\nLine two")        # \n = new line
print("Name:\tPriya")              # \t = tab
print("She said \"hello\"")        # \" = a quote mark inside quotes
````

````python
# A real use: pulling a date apart with slicing.
date = "2026-08-28"
print(f"Year:  {date[:4]}")
print(f"Month: {date[5:7]}")
print(f"Day:   {date[-2:]}")
````

# 8. Operators

> ⚠️ **`=` assigns. `==` compares.** One equals sign puts a value into a name; two ask a question. **The classic beginner bug.**

🧠 **`and` versus `or`.** To enter an exam hall you need a hall ticket **and** an ID card — miss either and you are out. To get a discount you might need to be a student **or** a senior citizen — either one is enough.

````python
age = 20            # ASSIGN 20 to age
print(age == 20)    # ASK: is age 20?  ->  True

marks = 78
print(marks > 40)         # True
print(marks == 100)       # False
print(marks != 78)        # False
print(type(marks > 40))   # <class 'bool'> - comparisons produce booleans
````

````python
# Logical operators combine conditions.
age = 20
has_id = True

print(age >= 18 and has_id)     # True  - BOTH hold
print(age >= 60 or has_id)      # True  - at least ONE holds
print(not has_id)               # False - flipped
````

````python
# Assignment shorthand, and membership.
total = 10
total += 5      # same as: total = total + 5
total -= 3
total *= 2
print(total)    # 24

text = "machine learning"
print("learn" in text)          # True
print("java" in text)           # False
print("java" not in text)       # True
````

````python
# Brackets matter more than you think.
is_student, is_senior, spend = True, False, 1200

print((is_student or is_senior) and spend > 1000)    # True
print(is_student or is_senior and spend > 1000)      # also True here, but
print()
print("...the second line means something DIFFERENT: 'and' binds tighter")
print("than 'or'. When it is not obvious, USE BRACKETS.")
````

# 9. Lists

🧠 **A shopping list on paper.** Items sit in a definite order. You can read the third one, cross one out, add one at the bottom, or change one.

> **Unlike strings, lists CAN be changed in place.** This is the most important difference between them.

````python
fruits = ["apple", "banana", "cherry", "date"]
#            0         1         2        3

print(fruits[0])       # apple
print(fruits[-1])      # date      - the last item
print(fruits[1:3])     # ['banana', 'cherry']   - stop still excluded
print(len(fruits))     # 4
````

````python
# Changing - this is what strings cannot do.
fruits = ["apple", "banana", "cherry"]
fruits[1] = "blueberry"
print(fruits)          # ['apple', 'blueberry', 'cherry']
````

````python
# Adding and removing.
fruits = ["apple", "banana"]
fruits.append("cherry")            # add to the END
fruits.insert(1, "apricot")        # insert AT position 1
print(fruits)

fruits.remove("banana")            # remove by VALUE (the first match)
last = fruits.pop()                # remove and RETURN the last item
print(f"popped {last}, left with {fruits}")
````

````python
# Statistics without writing any maths yourself.
marks = [78, 92, 45, 88, 61]

print(f"Count:   {len(marks)}")
print(f"Highest: {max(marks)}")
print(f"Lowest:  {min(marks)}")
print(f"Total:   {sum(marks)}")
print(f"Average: {sum(marks) / len(marks):.2f}")
print()
print(f"sorted() gives a NEW list: {sorted(marks)}")
print(f"the original is unchanged: {marks}")
marks.sort()
print(f".sort() changes it in place: {marks}")
````

# 10. Tuples and Sets

🧠 **Tuple — a printed certificate versus a whiteboard.** You can rewrite a whiteboard (a list). A printed certificate is fixed. **You use a tuple to say "this must not change".**

🧠 **Set — a guest list where each name appears once.** Adding a name already there changes nothing. **And nobody is "third on the list" — a set has no order at all.**

````python
colours = ("red", "green", "blue")      # round brackets = tuple

print(colours[0])        # red      - reading works just like a list
print(len(colours))      # 3
print("red" in colours)  # True

# colours[0] = "yellow"
# -> TypeError: 'tuple' object does not support item assignment
#    That refusal is the POINT of a tuple.

# Unpacking is where tuples really shine:
point = (3, 7)
x, y = point
print(f"x={x}, y={y}")
````

````python
numbers = {1, 2, 3, 2, 1}        # curly brackets = set
print(numbers)                    # {1, 2, 3} - duplicates gone automatically

s = {1, 2, 3}
s.add(4)
s.add(2)                          # unchanged - already there
s.discard(1)
s.discard(99)                     # discard never complains if missing
print(s)                          # {2, 3, 4}

# s[0]  -> TypeError: a set has no order, so there is no "first" item
````

````python
# Set maths - what makes sets worth learning.
maths   = {"Arun", "Priya", "Ravi"}
physics = {"Priya", "Ravi", "Meera"}

print(f"in EITHER class : {maths | physics}")     # union
print(f"in BOTH classes : {maths & physics}")     # intersection
print(f"maths only      : {maths - physics}")     # difference
print()

# The most common real use: removing duplicates.
visits = ["A101", "B202", "A101", "C303", "B202", "A101"]
print(f"{len(visits)} visits by {len(set(visits))} different people")
````

# 11. Dictionaries

🧠 **An actual dictionary.** You look up a *word* and get its *meaning*. You do not ask for "the 47th word". **You look things up by key, not by position.**

# 12. Conditionals: if, elif, else and match

🧠 **A signboard at a road junction.** *If the road is clear, go straight. Otherwise if there is a diversion, turn left. Otherwise, wait.* **Exactly one of those happens.**

> ⚠️ **Order matters enormously.** Python runs the **first** true branch and skips the rest. Always write the strictest condition first.

````python
marks = 84

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
elif marks >= 40:
    grade = "D"
else:
    grade = "Fail"

print(f"Marks {marks} -> Grade {grade}")
````

````python
# Why the ORDER matters. Here is the SAME ladder written backwards:
marks = 95

if marks >= 40:
    wrong = "D"
elif marks >= 90:
    wrong = "A"
else:
    wrong = "Fail"

print(f"Marks {marks} -> Grade {wrong}   <- WRONG!")
print()
print("95 is caught by the FIRST condition (>= 40) and Python stops there.")
print("The 'A' branch is never even looked at. Strictest condition FIRST.")
````

````python
# Combining conditions with and / or / not.
age, has_id = 20, True

if age >= 18 and has_id:
    print("Entry allowed")
else:
    print("Entry refused")

# The one-line form, for simple cases:
status = "adult" if age >= 18 else "minor"
print(status)
````

````python
# match - better than a chain of elifs when comparing ONE value
# against several fixed options. (Python 3.10 and later.)
choice = "2"

match choice:
    case "1":
        print("You chose Tea")
    case "2":
        print("You chose Coffee")
    case "3":
        print("You chose Samosa")
    case _:                       # _ means "anything else"
        print("Not on the menu")
````

## ⭐ Checkpoint Problem 3 — Even or odd

Read a whole number and say whether it is even or odd, and whether it is positive, negative or zero.

````python
# number = int(input("Enter a whole number: "))    # <- uncomment to try it
number = -7

# % gives the remainder. An even number divided by 2 leaves nothing over.
if number % 2 == 0:
    print(f"{number} is EVEN")
else:
    print(f"{number} is ODD")

if number > 0:
    print("It is positive")
elif number < 0:
    print("It is negative")
else:
    print("It is zero")

print()
print("Why % 2 == 0 and not / 2? Because / gives 3.5 or 4.0 - both are")
print("truthy, so they tell you nothing. % gives exactly 0 or 1, which")
print("is precisely the question you are asking.")
````

````python
# The harder extension: leap years. The ORDER is the whole puzzle.
for year in [2000, 1900, 2024, 2023]:
    if year % 400 == 0:
        result = "IS"
    elif year % 100 == 0:
        result = "is NOT"
    elif year % 4 == 0:
        result = "IS"
    else:
        result = "is NOT"
    print(f"{year} {result} a leap year")

print()
print("Checking 400 first, then 100, then 4 is what makes the rule work.")
print("Reverse them and 2000 gets the wrong answer.")
````

# 13. Loops: for and while

🧠 **Two kinds of repeated instruction.**
- *"Greet every guest on this list"* — you know how many. **That is a `for` loop.**
- *"Keep stirring until it thickens"* — you do not. **That is a `while` loop.**

````python
# range() for counting.
for i in range(5):
    print(i, end=" ")            # 0 1 2 3 4 - starts at 0, STOPS BEFORE 5
print()

for i in range(1, 6):
    print(i, end=" ")            # 1 2 3 4 5
print()

for i in range(0, 10, 2):
    print(i, end=" ")            # 0 2 4 6 8 - step of 2
print()

for i in range(5, 0, -1):
    print(i, end=" ")            # 5 4 3 2 1 - counting down
print()
````

````python
# while - repeat while a condition holds.
balance = 1000
year = 0

while balance < 2000:
    balance = balance * 1.10       # 10% growth each year
    year += 1                      # WITHOUT this, the loop never ends

print(f"It took {year} years to double, reaching {balance:.2f}")
````

````python
# break leaves the loop entirely. continue skips just this turn.
for i in range(1, 10):
    if i == 5:
        break
    print(i, end=" ")            # 1 2 3 4
print()

for i in range(1, 6):
    if i == 3:
        continue
    print(i, end=" ")            # 1 2 4 5
print()
````

````python
# THE pattern you will use constantly: build up a result.
marks = [78, 32, 92, 45, 38, 61]

total = 0                        # start with an empty result
passed = 0
for m in marks:
    total += m                   # add each item to it
    if m >= 40:
        passed += 1

print(f"Total:   {total}")
print(f"Average: {total / len(marks):.2f}")
print(f"Passed:  {passed} of {len(marks)}")
````

````python
# enumerate() when you need the position too.
tasks = ["wash", "cook", "study"]

for number, task in enumerate(tasks, start=1):
    print(f"{number}. {task}")
````

# 14. Functions

🧠 **A recipe you have written down.** You write it once, then say *"make chai"* rather than repeating the steps. **And if you improve the recipe, every future cup improves — you fix it in one place.**

> ⚠️ **`print` shows a value; `return` hands it back.** A function that prints cannot have its answer used in a calculation. A function that returns can.

````python
def greet(name):
    print(f"Hello, {name}!")

greet("Priya")
greet("Arun")
````

````python
# The print vs return difference, demonstrated.
def show(a, b):
    print(a + b)          # displays it, gives back None

def give(a, b):
    return a + b          # hands the value back

total = give(2, 3) * 10   # 50 - works
print(total)

print(show(2, 3))         # prints 5, then prints None - THAT is what it gave back
# total = show(2, 3) * 10 # TypeError: NoneType * int. Uncomment to see it.
````

````python
# Default values and keyword arguments.
def apply_discount(price, percent=10):
    """Return the price after a percentage discount."""
    return price * (1 - percent / 100)

print(apply_discount(1000))          # 900.0  - uses the default
print(apply_discount(1000, 25))      # 750.0
print(apply_discount(price=1000, percent=5))   # by name - order does not matter
````

## ⭐ Checkpoint Problem 5 — Prime number checker

````python
def is_prime(n):
    """Return True if n is a prime number."""
    if n < 2:                    # 0, 1 and negatives are not prime
        return False
    for divisor in range(2, n):
        if n % divisor == 0:     # it divided evenly, so it is NOT prime
            return False         # stop immediately - the answer is settled
    return True                  # nothing divided it: it is prime


for number in range(1, 50):
    if is_prime(number):
        print(number, end=" ")
print()
print()
print("return False is INSIDE the loop - once you find one divisor the")
print("answer is settled. return True is OUTSIDE - it only runs if the")
print("loop finished without ever finding one.")
````

````python
# The faster version: you only need divisors up to the square root.
def is_prime_fast(n):
    """Return True if n is prime. Only tests divisors up to sqrt(n)."""
    if n < 2:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    # If n has a divisor ABOVE its square root, it must also have a matching
    # one BELOW - which we would have found already. So stop at sqrt(n).
    for divisor in range(3, int(n ** 0.5) + 1, 2):
        if n % divisor == 0:
            return False
    return True

count = 0
for n in range(2, 1000):
    if is_prime_fast(n):
        count += 1
print(f"There are {count} primes below 1000")
````

# 15. Classes and Objects

🧠 **A cookie cutter and the cookies.** The cutter is the **class** — one design. Each cookie is an **object** — same shape, its own icing. **You define the class once; you make as many objects as you like.**

- **`__init__`** runs automatically when an object is created.
- **`self`** is the object being worked on. Python passes it in; you never do.
- **Attributes have no brackets; methods do.**

````python
class Student:
    def __init__(self, name, marks):
        self.name = name          # store on THIS object
        self.marks = marks

    def grade(self):              # a METHOD - note self comes first
        if self.marks >= 90:
            return "A"
        elif self.marks >= 75:
            return "B"
        elif self.marks >= 40:
            return "C"
        return "F"


arun = Student("Arun", 78)
priya = Student("Priya", 92)

print(arun.name)       # attribute - no brackets
print(arun.grade())    # method    - brackets
print(priya.grade())

arun.marks = 85        # attributes can be changed
print(arun.grade())    # recalculated from the new value
````

````python
# __str__ decides how the object prints.
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price

    def __str__(self):
        return f"'{self.title}' by {self.author} - {self.price}"


print(Book("Python Basics", "R. Nair", 450))
````

````python
# A complete, useful class: the data and the rules that protect it,
# living together.
class BankAccount:
    def __init__(self, holder, balance=0):
        self.holder = holder
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            return "Deposit must be positive"
        self.balance += amount
        return f"Deposited {amount}. Balance: {self.balance}"

    def withdraw(self, amount):
        if amount > self.balance:
            return "Insufficient funds"
        self.balance -= amount
        return f"Withdrew {amount}. Balance: {self.balance}"

    def __str__(self):
        return f"{self.holder}: {self.balance}"


account = BankAccount("Priya", 1000)
print(account.deposit(500))
print(account.withdraw(2000))       # refused
print(account.withdraw(400))
print(account)

print()
print("Nothing OUTSIDE the class can withdraw more than the balance,")
print("because the check lives INSIDE withdraw(). That is the point.")
````

````python
# Inheritance: one class built on another.
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        return f"I am {self.name}, {self.age} years old"


class Student2(Person):                  # Student2 INHERITS from Person
    def __init__(self, name, age, course):
        super().__init__(name, age)      # run Person's __init__ first
        self.course = course

    def introduce(self):                 # replace the inherited version
        return f"{super().introduce()} and I study {self.course}"


print(Person("Ravi", 45).introduce())
print(Student2("Priya", 20, "B.Tech").introduce())
````

### Why this matters for the rest of the course

**Every scikit-learn model you meet from Session 4 onwards is an object of a class.**

```python
# This is Session 4 code - read it for the SHAPE.
model = RandomForestClassifier(n_estimators=100)   # __init__ runs
model.fit(X_train, y_train)                        # a method
model.predict(X_test)                              # another method
model.feature_importances_                         # an attribute
```

**You are doing exactly what you just did with `BankAccount`.** Here is that shape in miniature:

````python
class SimpleAverager:
    """Predicts the average of whatever it was trained on."""

    def __init__(self):
        self.prediction = None            # nothing learned yet

    def fit(self, y):
        """Learn from the data."""
        self.prediction = sum(y) / len(y)
        return self                       # so you can chain .fit().predict()

    def predict(self, n):
        """Predict for n new items."""
        if self.prediction is None:
            raise ValueError("Call fit() before predict()")
        return [self.prediction] * n


model = SimpleAverager()
model.fit([10, 20, 30, 40])
print(model.predict(3))         # [25.0, 25.0, 25.0]

print()
print("That is the ENTIRE scikit-learn pattern: create with settings,")
print("fit to learn, predict to use. Every model in Sessions 4 to 8")
print("follows it.")
````
